```python
import random

SURNAME = "Pena"
SEED_NUM = 8

random.seed(SEED_NUM)

NUM_READINGS = 10

OPERATING_RANGES = {
    "Normal": (40, 60),
    "Low": (0, 39.99),
    "High": (60.01, 100)
}

readings = [
    round(random.uniform(0, 100), 2)
    for _ in range(NUM_READINGS)
]

print("Sensor Readings")
print("=" * 30)

for i, reading in enumerate(readings, start=1):
    print(f"Reading {i}: {reading}")


def classify_reading(value):
    for status, (lower, upper) in OPERATING_RANGES.items():
        if lower <= value <= upper:
            return status
    return "Out of Range"


print("\nClassification")
print("=" * 30)

summary = {}

for i, reading in enumerate(readings, start=1):
    status = classify_reading(reading)
    print(f"Reading {i}: {reading} -> {status}")
    summary[status] = summary.get(status, 0) + 1


print("\nSummary")
print("=" * 30)

for status, count in summary.items():
    print(f"{status}: {count}")

print(f"\nStudent: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")
```

```python
import random
import string

SURNAME = "Pena"
SEED_NUM = 8

# Use surname and seed number to create a unique signal
seed_value = sum(ord(char) for char in SURNAME) + SEED_NUM
random.seed(seed_value)

characters = string.ascii_letters + string.digits + " !@#$%^&*_-"

signal = ''.join(random.choice(characters) for _ in range(20))

# Process and normalize the signal
processed_signal = signal.lower().strip()

print("Generated Signal:")
print(signal)

print("\nProcessed Signal:")
print(processed_signal)

# Character Analysis
letters = 0
numbers = 0
spaces = 0
others = 0

print("\nCharacter Analysis:")

for char in processed_signal:
    if char.isalpha():
        classification = "Letter"
        letters += 1
    elif char.isdigit():
        classification = "Number"
        numbers += 1
    elif char.isspace():
        classification = "Space"
        spaces += 1
    else:
        classification = "Other Character"
        others += 1

    print(f"{repr(char)} -> {classification}")

# Final classification
if letters > 0 and numbers > 0 and others > 0:
    signal_classification = "Mixed Character Signal"
elif letters > 0 and numbers > 0:
    signal_classification = "Alphanumeric Signal"
elif letters > 0:
    signal_classification = "Letter Signal"
elif numbers > 0:
    signal_classification = "Numeric Signal"
else:
    signal_classification = "Special Character Signal"

print("\nSignal Classification:")
print(signal_classification)

print("\nExecution Log:")
print(f"Student: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"Signal Length: {len(processed_signal)}")
print("Character-by-character analysis completed.")

print("\nFinal Output:")
print(f"Student: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"Classification: {signal_classification}")
print(f"Letters: {letters}")
print(f"Numbers: {numbers}")
print(f"Spaces: {spaces}")
print(f"Other Characters: {others}")
```

```python
import random
import string

SURNAME = "Pena"
SEED_NUM = 8

# Generate a unique password using surname and SEED_NUM
seed_value = sum(ord(char) for char in SURNAME) + SEED_NUM
random.seed(seed_value)

password_characters = string.ascii_letters + string.digits
generated_password = ''.join(
    random.choice(password_characters) for _ in range(8)
)

# Generate the allowed attempt limit
attempt_limit = 3 + (SEED_NUM % 3)

# Normalize the password
normalized_password = generated_password.strip()

print("Generated Password:")
print(normalized_password)

print("\nAttempt Limit:")
print(attempt_limit)

# Authentication simulation
attempts_made = 0
access_granted = False

# Use the correct generated password for the final attempt
attempts = ["wrong123", "incorrect", "test123", normalized_password]

for attempt in attempts:
    if attempts_made >= attempt_limit:
        break

    attempts_made += 1

    if attempt == normalized_password:
        access_granted = True
        print(f"Attempt {attempts_made}: Access Granted")
        break
    else:
        print(f"Attempt {attempts_made}: Access Denied")

# Final system state
if access_granted:
    system_state = "UNLOCKED"
    access_result = "Access Granted"
else:
    system_state = "LOCKED"
    access_result = "Access Denied"

print("\nFinal System State:")
print(system_state)

print("\nExecution Log:")
print(f"Student: {SURNAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"Attempts Made: {attempts_made}")
print(f"Attempt Limit: {attempt_limit}")
print(f"Access Result: {access_result}")
print(f"Final System State: {system_state}")

print("\nFinal Output:")
print(f"Generated Password: {normalized_password}")
print(f"Attempt Limit: {attempt_limit}")
print(f"Attempts Made: {attempts_made}")
print(f"Access Result: {access_result}")
print(f"Final System State: {system_state}")
```
